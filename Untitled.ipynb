### 求每个产品的单价和成本

```python
# 定义函数
def myfunc(df):
    units=df["销售额"]/(df["折扣"]*df["数量"])#单价
    units=units.dropna().values[0]#返回每行单价去除缺失值后的第一个值作为产品的单价
    costs=(df["销售额"]-df["利润"])/df["数量"]#成本
    costs=costs.dropna().values[0]
    return pd.DataFrame({"单价":[units],"成本":[costs]})

product_table=data.groupby(by="产品 ID").apply(myfunc).reset_index().drop(labels="level_1",axis=1)#对产品进行分组，使用myfunc函数计算每个产品的单价，将结果作为产品表
#product_table.rename(columns={0:"单价"},inplace=True)#修改列名为单价
```

```python
data=pd.merge(left=data,right=product_table,how="left",on="产品 ID")#两表连接
```

```python
data
```

### 填补四列缺失

```python
#填补销售额
for idx in data.loc[data["销售额"].isna(),].index:
    data.loc[idx,"销售额"]=data.loc[idx,"数量"]*data.loc[idx,"折扣"]*data.loc[idx,"单价"]
```

```python
data.loc[data["销售额"].isna(),]
```

```python
#填补数量
for idx in data.loc[data["数量"].isna(),].index:
    data.loc[idx,"数量"]=data.loc[idx,"销售额"]/(data.loc[idx,"单价"]*data.loc[idx,"折扣"])
```

```python
data.loc[data["数量"].isna(),]
```

```python
#填补折扣
for idx in data.loc[data["折扣"].isna(),].index:
    data.loc[idx,"折扣"]=data.loc[idx,"销售额"]/(data.loc[idx,"单价"]*data.loc[idx,"数量"])
```

```python
data.loc[data["折扣"].isna(),]
```

```python
#填补利润
for idx in data.loc[data["利润"].isna(),].index:
    data.loc[idx,"利润"]=((data.loc[idx,"单价"]*data.loc[idx,"折扣"])-data.loc[idx,"成本"])*data.loc[idx,"数量"]
```

# 决策分析

## 产品波斯顿矩阵分析

```python
#计算每个产品今年的销售额占比，同比增长率
def myfunc(df):
    if df["订单日期"].dt.year.nunique()==2:
        all_gmv=data[data["订单日期"].dt.year==2023]["销售额"].sum()
        gmv=df[df["订单日期"].dt.year==2023]["销售额"].sum()
        gmv_rate=gmv/all_gmv
        gmv_year=df.groupby(df["订单日期"].dt.year)["销售额"].sum()
        gmv_yoy=(gmv_year.values[-1]-gmv_year.values[-2])/gmv_year.values[-2]
        return pd.DataFrame({"销售额占有率":[gmv_rate],"销售额增长率":[gmv_yoy]})
    else:
        return pd.DataFrame({"销售额占有率":[0],"销售额增长率":[0]})
    
result=data[data["订单日期"].dt.year.isin([2022,2023])].groupby("产品 ID").apply(myfunc).reset_index().drop("level_1",axis=1)
result=result[result["销售额占有率"]>0]
#计算销售额占比均值，同比增长率均值,使用均值划分产品为四类
def get_bcg(x):
    if(x["销售额占有率"]>=result["销售额占有率"].mean())&(x["销售额增长率"]>=result["销售额增长率"].mean()):
        return "明星产品"
    elif (x["销售额占有率"]<result["销售额占有率"].mean())&(x["销售额增长率"]>=result["销售额增长率"].mean()):
        return "问题产品"
    elif (x["销售额占有率"]<result["销售额占有率"].mean())&(x["销售额增长率"]<result["销售额增长率"].mean()):
        return "瘦狗产品"
    else:
        return "金牛产品"
result["BCG分类"]=result.apply(get_bcg,axis=1)
#绘制散点图展示波士顿分析
plt.figure(figsize=(5,5))
sns.scatterplot(data=result,x="销售额占有率",y="销售额增长率",hue="BCG分类")
plt.title("产品波士顿矩阵分析")
plt.show()
```

## 产品帕累托模型

```python
#统计今年每个产品的销售额
result=data[data["订单日期"].dt.year==2023].groupby("产品 ID")["利润"].sum().reset_index().sort_values("利润",ascending=False)
#销售额占比
result["累计利润"]=result["利润"].cumsum()
result["累计占比"]=result["累计利润"]/result["利润"].sum()
#划分ABC分类
def myfunc(x):
    if x<=0.8:
        return "A类产品"
    elif x<=0.9:
        return "B类产品"
    else:
        return "C类产品"
result["ABC分类"]=result["累计占比"].apply(myfunc)
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=result,x="产品 ID",y="利润",hue="ABC分类")
plt.show()
```

## RFM用户画像

```python
# 计算每个客户的最近一次消费时间距离最后统计日期的时间间隔day、消费次数、消费金额
def myfunc(df):
    R_value=data["订单日期"].max()-df["订单日期"].max()
    F_value=df["订单 ID"].nunique()
    M_value=(df["数量"]*df["折扣"]*df["单价"]).sum()
    return pd.DataFrame({"R_value":[R_value.days],"F_value":[F_value],"M_value":[M_value]})

result=data.groupby(by="客户 ID").apply(myfunc).reset_index().drop("level_1",axis=1)
```

```python
def get_RFM(x):
    if (x["R_value"]>=result["R_value"].mean())&(x["F_value"]>=result["F_value"].mean())&(x["M_value"]>=result["M_value"].mean()):
        return "重要保持客户"
    elif (x["R_value"]<result["R_value"].mean())&(x["F_value"]>=result["F_value"].mean())&(x["M_value"]>=result["M_value"].mean()):
        return "重要价值客户"
    elif (x["R_value"]>=result["R_value"].mean())&(x["F_value"]<result["F_value"].mean())&(x["M_value"]>=result["M_value"].mean()):
        return "重要挽留客户"
    elif (x["R_value"]>=result["R_value"].mean())&(x["F_value"]>=result["F_value"].mean())&(x["M_value"]<result["M_value"].mean()):
        return "一般保持客户"
    elif (x["R_value"]<result["R_value"].mean())&(x["F_value"]<result["F_value"].mean())&(x["M_value"]>=result["M_value"].mean()):
        return "重要发展客户"
    elif (x["R_value"]<result["R_value"].mean())&(x["F_value"]>=result["F_value"].mean())&(x["M_value"]<result["M_value"].mean()):
        return "一般价值客户"
    elif (x["R_value"]>=result["R_value"].mean())&(x["F_value"]<result["F_value"].mean())&(x["M_value"]<result["M_value"].mean()):
        return "一般潜在客户"
    else:
        return "一般发展客户"
result["RFM分类"]=result.apply(get_RFM,axis=1)
```
